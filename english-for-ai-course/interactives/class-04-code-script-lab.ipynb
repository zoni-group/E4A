## Step 2. Plan a safe project

Choose one project. Change the example text, then run the cell.

````python
# @title Step 2: Plan a safe project
project_type = "Guess the Number game"
game_goal = "The player guesses a secret number from 1 to 100."
feature_one = "The game says Too high, Too low, or Correct."
feature_two = "The game counts attempts and has a New Game button."
feature_three = "The game rejects empty and out-of-range input."

show(
    "**My project:** " + project_type + "\n\n"
    "**Goal:** " + game_goal + "\n\n"
    "**Features:**\n"
    "- " + feature_one + "\n"
    "- " + feature_two + "\n"
    "- " + feature_three
)
````

## Step 3. Build the complete-HTML prompt

The output below is a **prompt plan**, not game code. The complete-HTML
rule is included every time.

````python
# @title Step 3: Build the complete-HTML prompt
complete_html_rule = (
    "Create one complete, self-contained HTML document.\n\n"
    "Put all CSS inside a <style> element.\n"
    "Put all JavaScript inside a <script> element.\n"
    "Do not use external libraries, images, fonts, audio, or network requests.\n"
    "The document must run in a browser.\n\n"
    "Return the entire HTML document in one code block.\n"
    "Start with <!doctype html>.\n"
    "Do not return partial code or only the changed section."
)

complete_prompt = (
    "I want to create a " + project_type + ".\n\n"
    + "Goal: " + game_goal + "\n"
    + "Required feature 1: " + feature_one + "\n"
    + "Required feature 2: " + feature_two + "\n"
    + "Required feature 3: " + feature_three + "\n\n"
    + complete_html_rule
)

show("**My complete-HTML prompt:**\n\n```text\n" + complete_prompt + "\n```")
````

## Step 4. Plan one test and bug report

Write what should happen before asking AI to change anything.

````python
# @title Step 4: Plan one test and bug report
test_action = "Leave the number input empty and select Guess."
expected_result = "The game shows a helpful message and does not add an attempt."
actual_result = "The game shows no message and adds one attempt."
bug_sentence = "The bug happens when the input is empty."

debug_prompt = (
    "Here is my current complete HTML document.\n\n"
    "I expected: " + expected_result + "\n"
    "The actual result was: " + actual_result + "\n"
    + bug_sentence + "\n\n"
    "Find the bug and correct it.\n"
    "Keep all existing working features.\n"
    "Return the complete corrected HTML document, not a fragment."
)

show(
    "**Test action:** " + test_action + "\n\n"
    "**Expected result:** " + expected_result + "\n\n"
    "**Actual result:** " + actual_result + "\n\n"
    "**Debugging request:**\n\n```text\n" + debug_prompt + "\n```"
)
````
